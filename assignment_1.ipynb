```python

!pip -q install web3

from web3 import Web3
from eth_account import Account
from getpass import getpass

print("=" * 45)
print("          HELLO BLOCKCHAIN")
print("=" * 45)

# STEP 1: Connect to Ethereum blockchain
print("\nSTEP 1: BLOCKCHAIN CONNECTION")

RPC_URL = "https://ethereum-rpc.publicnode.com"
w3 = Web3(Web3.HTTPProvider(RPC_URL, request_kwargs={"timeout": 15}))

if w3.is_connected():
    print("Blockchain Connection: Successful")
    print("Network is ready for interaction.")
    print("Chain ID:", w3.eth.chain_id)
else:
    print("Blockchain Connection: Failed")
    print("Check your internet connection or RPC endpoint.")

# STEP 2: Create a test wallet
print("\nSTEP 2: CREATE A TEST WALLET")

new_account = Account.create()

print("Wallet Created Successfully")
print("Wallet Address:", new_account.address)
print("Private Key:", new_account.key.hex())
print("WARNING: Never share a real private key.")
print("This is a newly generated disposable wallet.")

# STEP 3: Load a wallet using a private key
print("\nSTEP 3: LOAD A WALLET")

use_new = input(
    "Load the newly created wallet? (yes/no): "
).strip().lower()

if use_new == "yes":
    loaded_account = Account.from_key(new_account.key)
else:
    private_key = getpass(
        "Enter a disposable test wallet private key: "
    ).strip()

    loaded_account = Account.from_key(private_key)

print("Wallet Loaded Successfully")
print("Wallet Address:", loaded_account.address)

# STEP 4: Check wallet balance
print("\nSTEP 4: CHECK WALLET BALANCE")

if w3.is_connected():
    balance_wei = w3.eth.get_balance(loaded_account.address)
    balance_eth = Web3.from_wei(balance_wei, "ether")

    print("Wallet Balance")
    print("-" * 30)
    print("Balance in Wei :", balance_wei)
    print("Balance in ETH :", f"{balance_eth:.6f}")
else:
    print("Balance cannot be retrieved.")
    print("The blockchain connection is unavailable.")

print("\nPRACTICAL COMPLETED")
```
