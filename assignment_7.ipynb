```python

import os
from IPython.display import display, HTML

os.makedirs("/content/PortfolioDApp", exist_ok=True)

html = r'''<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>Personal Portfolio dApp</title>

<script src="https://cdn.jsdelivr.net/npm/web3@4.16.0/dist/web3.min.js"></script>

<style>
* {
    box-sizing: border-box;
    margin: 0;
    padding: 0;
}

body {
    font-family: Arial, sans-serif;
    background: linear-gradient(135deg,#0f172a,#172554);
    color: white;
    min-height: 100vh;
}

header {
    text-align: center;
    padding: 30px 15px;
    background: #020617;
}

header h1 {
    color: #38bdf8;
    font-size: clamp(24px,5vw,36px);
}

header p {
    color: #cbd5e1;
    margin-top: 10px;
}

.container {
    width: min(90%,650px);
    margin: 55px auto;
    padding: clamp(20px,5vw,40px);
    background: #1e293b;
    border-radius: 18px;
    box-shadow: 0 10px 35px #0006;
    text-align: center;
}

.container h2 {
    margin-bottom: 20px;
}

button {
    display: inline-block;
    padding: 14px 22px;
    margin: 10px 5px;
    border: none;
    border-radius: 9px;
    background: #0284c7;
    color: white;
    font-size: 15px;
    font-weight: bold;
    cursor: pointer;
    transition: .2s;
    max-width: 100%;
}

button:hover {
    background: #0369a1;
    transform: translateY(-2px);
}

button:disabled {
    opacity: .6;
    cursor: wait;
}

.profile {
    margin: 25px 0;
    padding: 30px 15px;
    background: #334155;
    border-radius: 15px;
}

.avatar {
    width: 85px;
    height: 85px;
    border-radius: 50%;
    background: #0ea5e9;
    display: grid;
    place-items: center;
    font-size: 35px;
    margin: 0 auto 20px;
}

.profile h2 {
    color: #38bdf8;
    font-size: 27px;
}

.profile h3 {
    margin-bottom: 15px;
    color: #cbd5e1;
}

.skills {
    line-height: 1.8;
    overflow-wrap: anywhere;
}

#wallet, #status {
    margin: 15px 0;
    color: #7dd3fc;
    overflow-wrap: anywhere;
    line-height: 1.6;
}

footer {
    text-align: center;
    padding: 25px;
    color: #94a3b8;
}

@media(max-width:480px) {
    .container {
        margin: 25px auto;
    }

    button {
        width: 100%;
        margin: 8px 0;
    }

    .profile h2 {
        font-size: 23px;
    }
}
</style>
</head>

<body>

<header>
    <h1>Personal Portfolio dApp</h1>
    <p>Decentralized Portfolio Application</p>
</header>

<div class="container">

    <h2>Welcome to My Portfolio</h2>

    <button id="connectBtn" onclick="connectWallet()">
        Connect MetaMask
    </button>

    <p id="wallet">Wallet: Not Connected</p>

    <div class="profile">

        <div class="avatar">M</div>

        <h2 id="name">Manikanta</h2>

        <h3 id="role">B.Tech CSE Student</h3>

        <p class="skills" id="skills">
            Python | HTML | CSS | JavaScript | Blockchain
        </p>

    </div>

    <button id="loadBtn" onclick="loadPortfolio()">
        Load Portfolio
    </button>

    <p id="status">Demo portfolio is ready.</p>

</div>

<footer>
    Blockchain Engineering | Personal Portfolio dApp
</footer>

<script>

let web3;
let contract;
let account = null;

const contractAddress = "YOUR_DEPLOYED_CONTRACT_ADDRESS";

const contractABI = [
    {
        "inputs": [],
        "name": "getPortfolio",
        "outputs": [
            {"internalType":"string","name":"","type":"string"},
            {"internalType":"string","name":"","type":"string"},
            {"internalType":"string","name":"","type":"string"}
        ],
        "stateMutability":"view",
        "type":"function"
    }
];

async function connectWallet() {

    const status = document.getElementById("status");
    const wallet = document.getElementById("wallet");
    const button = document.getElementById("connectBtn");

    if (!window.ethereum) {
        status.innerText =
            "MetaMask not detected. Open this page in Chrome with MetaMask installed.";
        return;
    }

    try {
        button.disabled = true;
        status.innerText = "Connecting wallet...";

        const accounts = await window.ethereum.request({
            method: "eth_requestAccounts"
        });

        account = accounts[0];
        web3 = new Web3(window.ethereum);

        wallet.innerText = "Connected: " + account;

        if (contractAddress !== "YOUR_DEPLOYED_CONTRACT_ADDRESS") {
            contract = new web3.eth.Contract(
                contractABI,
                contractAddress
            );
        }

        status.innerText = "MetaMask connected successfully!";

    } catch (error) {
        status.innerText = "Connection failed: " + error.message;
    } finally {
        button.disabled = false;
    }
}

async function loadPortfolio() {

    const status = document.getElementById("status");
    const button = document.getElementById("loadBtn");

    try {
        button.disabled = true;
        status.innerText = "Loading portfolio...";

        if (contract) {

            const result = await contract.methods.getPortfolio().call();

            document.getElementById("name").innerText = result[0];
            document.getElementById("role").innerText = result[1];
            document.getElementById("skills").innerText = result[2];

            status.innerText =
                "Portfolio loaded successfully from blockchain.";

        } else {

            document.getElementById("name").innerText = "Manikanta";
            document.getElementById("role").innerText =
                "B.Tech CSE Student";

            document.getElementById("skills").innerText =
                "Python | HTML | CSS | JavaScript | Blockchain";

            status.innerText =
                "Demo portfolio loaded. Deploy the contract and add its address for blockchain data.";
        }

    } catch (error) {
        status.innerText = "Error: " + error.message;
    } finally {
        button.disabled = false;
    }
}

if (window.ethereum) {
    window.ethereum.on("accountsChanged", function(accounts) {
        account = accounts[0] || null;

        document.getElementById("wallet").innerText =
            account ? "Connected: " + account : "Wallet: Not Connected";

        if (!account) {
            contract = null;
        }
    });
}

</script>

</body>
</html>'''

path = "/content/PortfolioDApp/index.html"

with open(path, "w", encoding="utf-8") as f:
    f.write(html)

# Open the webpage through a separate browser tab
from google.colab import output

display(HTML("""
<div style="padding:20px;text-align:center">
<h3>Personal Portfolio dApp is ready</h3>
<button onclick="google.colab.kernel.invokeFunction(
'open_portfolio', [], {}
)"
style="padding:14px 25px;background:#0284c7;color:white;
border:0;border-radius:8px;cursor:pointer;font-size:16px">
Open Portfolio Webpage
</button>
</div>
"""))

def open_portfolio():
    from google.colab import files
    files.download(path)

output.register_callback("open_portfolio", open_portfolio)

print("Responsive portfolio webpage created successfully!")
print("Use the Open Portfolio Webpage button to download index.html.")
```
