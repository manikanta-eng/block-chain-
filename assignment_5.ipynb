```python

%%writefile /content/PersonalPortfolio/contracts/PersonalPortfolio.sol
// SPDX-License-Identifier: MIT
pragma solidity ^0.8.20;

contract PersonalPortfolio {
    string public name;
    string public skills;
    string public description;
    address public owner;

    constructor() {
        owner = msg.sender;
    }

    function setProfile(
        string memory _name,
        string memory _skills,
        string memory _description
    ) public {
        name = _name;
        skills = _skills;
        description = _description;
    }

    function getProfile()
        public
        view
        returns (string memory, string memory, string memory)
    {
        return (name, skills, description);
    }
}
```

```python

%%writefile /content/PersonalPortfolio/truffle-config.js
module.exports = {
  networks: {},
  compilers: {
    solc: {
      version: "0.8.20",
      settings: {
        optimizer: {
          enabled: true,
          runs: 200
        }
      }
    }
  }
};
```

```python

%%writefile /content/PersonalPortfolio/migrations/1_deploy_contract.js
const PersonalPortfolio = artifacts.require("PersonalPortfolio");

module.exports = function (deployer) {
  deployer.deploy(PersonalPortfolio);
};
```

```python

%%writefile /content/PersonalPortfolio/test/personal_portfolio_test.js
const PersonalPortfolio = artifacts.require("PersonalPortfolio");

contract("PersonalPortfolio", (accounts) => {
  let portfolio;

  beforeEach(async () => {
    portfolio = await PersonalPortfolio.new();
  });

  it("should store the profile details", async () => {
    await portfolio.setProfile(
      "Manikanta",
      "Python, HTML, CSS, JavaScript",
      "Computer Science Student"
    );

    assert.equal(await portfolio.name(), "Manikanta");
    assert.equal(
      await portfolio.skills(),
      "Python, HTML, CSS, JavaScript"
    );
    assert.equal(
      await portfolio.description(),
      "Computer Science Student"
    );
  });

  it("should retrieve all profile details", async () => {
    await portfolio.setProfile(
      "Manikanta",
      "Python",
      "Developer"
    );

    const profile = await portfolio.getProfile();

    assert.equal(profile[0], "Manikanta");
    assert.equal(profile[1], "Python");
    assert.equal(profile[2], "Developer");
  });

  it("should update the profile details", async () => {
    await portfolio.setProfile("Manikanta", "Python", "Student");
    await portfolio.setProfile("Mani", "Solidity", "Blockchain Developer");

    assert.equal(await portfolio.name(), "Mani");
    assert.equal(await portfolio.skills(), "Solidity");
    assert.equal(
      await portfolio.description(),
      "Blockchain Developer"
    );
  });

  it("should record the contract deployer's address", async () => {
    assert.equal(await portfolio.owner(), accounts[0]);
  });
});
```
