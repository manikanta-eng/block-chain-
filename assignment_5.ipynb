```python

import os
import subprocess
import textwrap

os.chdir("/content")

def run(command):
    result = subprocess.run(
        command,
        shell=True,
        text=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT
    )
    print(result.stdout)
    if result.returncode != 0:
        raise RuntimeError("Command failed. Check the output above.")

print("STEP 1: Installing Node.js tools")
run("node --version && npm --version")

print("STEP 2: Creating project")
run("mkdir -p /content/PersonalPortfolio")
os.chdir("/content/PersonalPortfolio")

print("STEP 3: Initializing project")
run("npm init -y")

print("STEP 4: Installing Truffle and Ganache")
run("npm install --save-dev truffle@5.11.5 ganache@7.9.2")

os.makedirs("contracts", exist_ok=True)
os.makedirs("migrations", exist_ok=True)
os.makedirs("test", exist_ok=True)

contract_code = """
// SPDX-License-Identifier: MIT
pragma solidity ^0.8.20;

contract PersonalPortfolio {
    string public name;
    string public education;
    string public skills;
    string public email;

    function setProfile(
        string memory _name,
        string memory _education,
        string memory _skills,
        string memory _email
    ) public {
        name = _name;
        education = _education;
        skills = _skills;
        email = _email;
    }

    function getProfile()
        public
        view
        returns (string memory, string memory, string memory, string memory)
    {
        return (name, education, skills, email);
    }
}
"""

with open("contracts/PersonalPortfolio.sol", "w") as f:
    f.write(textwrap.dedent(contract_code))

migration_code = """
const PersonalPortfolio = artifacts.require("PersonalPortfolio");

module.exports = function (deployer) {
    deployer.deploy(PersonalPortfolio);
};
"""

with open("migrations/1_initial_migration.js", "w") as f:
    f.write(textwrap.dedent(migration_code))

test_code = """
const PersonalPortfolio = artifacts.require("PersonalPortfolio");

contract("PersonalPortfolio", () => {
    let portfolio;

    beforeEach(async () => {
        portfolio = await PersonalPortfolio.new();
    });

    it("stores and retrieves profile information", async () => {
        await portfolio.setProfile(
            "Manikanta",
            "B.Tech CSE",
            "Python, Solidity, JavaScript",
            "mani@example.com"
        );

        const profile = await portfolio.getProfile();

        assert.equal(profile[0], "Manikanta");
        assert.equal(profile[1], "B.Tech CSE");
        assert.equal(profile[2], "Python, Solidity, JavaScript");
        assert.equal(profile[3], "mani@example.com");
    });

    it("updates profile information", async () => {
        await portfolio.setProfile(
            "Manikanta", "B.Tech CSE", "Python", "mani@example.com"
        );

        await portfolio.setProfile(
            "Manikanta", "B.Tech CSE", "Python, Solidity",
            "mani@example.com"
        );

        const profile = await portfolio.getProfile();
        assert.equal(profile[2], "Python, Solidity");
    });

    it("initially contains empty profile fields", async () => {
        const profile = await portfolio.getProfile();

        assert.equal(profile[0], "");
        assert.equal(profile[1], "");
        assert.equal(profile[2], "");
        assert.equal(profile[3], "");
    });
});
"""

with open("test/personalPortfolio.test.js", "w") as f:
    f.write(textwrap.dedent(test_code))

config_code = """
module.exports = {
    networks: {
        development: {
            host: "127.0.0.1",
            port: 8545,
            network_id: "*"
        }
    },
    compilers: {
        solc: {
            version: "0.8.20"
        }
    }
};
"""

with open("truffle-config.js", "w") as f:
    f.write(textwrap.dedent(config_code))

print("STEP 5: Starting local blockchain")
ganache_log = open("/content/ganache.log", "w")
ganache = subprocess.Popen(
    [
        "./node_modules/.bin/ganache",
        "--server.host", "127.0.0.1",
        "--server.port", "8545",
        "--wallet.totalAccounts", "10"
    ],
    stdout=ganache_log,
    stderr=subprocess.STDOUT
)

import time
import socket

ready = False
for _ in range(30):
    try:
        with socket.create_connection(("127.0.0.1", 8545), timeout=1):
            ready = True
            break
    except OSError:
        time.sleep(1)

if not ready:
    ganache_log.flush()
    print(open("/content/ganache.log").read()[-5000:])
    raise RuntimeError("Ganache did not start.")

print("Ganache started successfully.")

print("STEP 6: Compiling smart contract")
run("npx truffle compile")

print("STEP 7: Deploying and running unit tests")
run("npx truffle test --network development")

print("STEP 8: Practical completed")
print("Smart contract compiled, deployed, and tested.")
```
